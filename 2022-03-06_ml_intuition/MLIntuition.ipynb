```python
import numpy as np

data = [
    (120, 500000),
    (55, 300000),
    (40,28000),
    (30, 250000)
]
```

```python
import matplotlib.pyplot as plt
ndata = np.array(data)
x=ndata[:,0]
y=ndata[:,1]

fig = plt.figure()
ax1 = fig.add_subplot(111)

def function_to_tune(a, b):
    return lambda x: a*x + b


predict_function = function_to_tune(5000,2)

x_p = list(range(min(x), max(x)+1))
y_p = [ predict_function(x_i) for x_i in x_p]



ax1.scatter(x, y, s=10,color='blue', label='real data')
ax1.scatter(x_p, y_p, s=10, color='red', label='prediction')
```

```python
plt.show()
```
